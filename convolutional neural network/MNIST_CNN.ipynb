# MNIST CNN Digit Recognizer

Run from the repository directory with dependencies installed. This notebook uses the same modules as the command-line application. Training downloads MNIST if it is not cached. No measured outputs are bundled.

```python
from pathlib import Path
assert Path("train.py").is_file(), "Change into the extracted repository directory first."
```

## Train, select by validation loss, and evaluate

The next cell runs up to 15 epochs. For a pipeline smoke run, use `--epochs 1`; this still uses the full dataset.

```python
import subprocess, sys
subprocess.run([sys.executable, "train.py", "--epochs", "15"], check=True)
```

## Inspect actual measured results

```python
import json
metrics = json.loads(Path("artifacts/metrics.json").read_text())
print(json.dumps(metrics, indent=2))
```

```python
from IPython.display import display, Image
for name in ["learning_curves.png", "confusion_matrix.png", "misclassified.png"]:
    path = Path("artifacts") / name
    if path.exists():
        display(Image(filename=str(path)))
```

## Predict one example

This demonstrates the external-image pipeline. Its preprocessing differs from direct MNIST benchmark evaluation.
